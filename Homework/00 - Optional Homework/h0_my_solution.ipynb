```python
# Download dataset
import pandas as pd

url = "https://raw.githubusercontent.com/fivethirtyeight/guns-data/master/interactive_data.csv"
data = pd.read_csv(url)

data.head()
```

```python
# Clean intent data
# clean_intent_data = data[data['Intent'].notna() & ~data['Intent'].isin(['None selected', 'Unknown'])]
clean_intent_data = data[data['Intent'] != 'None selected']
clean_intent_data.head()
```

```python
# Verify: Nearly two-thirds of gun deaths are suicides, on uncleaned data
death_count = data['Deaths'].sum()

suicides_df = data[data['Intent'] == 'Suicide']
suicides_count = suicides_df['Deaths'].sum()

suicides_count / death_count
```

```python
# Verify: Nearly two-thirds of gun deaths are suicides
death_count = clean_intent_data['Deaths'].sum()

suicides_df = clean_intent_data[clean_intent_data['Intent'] == 'Suicide']
suicides_count = suicides_df['Deaths'].sum()

suicides_count / death_count
```

```python
# Verify: Around a third of all gun deaths are homicides
death_count = clean_intent_data['Deaths'].sum()

homicides_df = clean_intent_data[clean_intent_data['Intent'] == 'Homicide']
homicides_count = homicides_df['Deaths'].sum()

homicides_count / death_count
```

```python
# Clean intent, gender data
clean_intent_gender_data = data[(data['Intent'] != 'None selected') & (data['Gender'] != 'None selected')]
clean_intent_gender_data.head()
```

```python
# Verify: More than 85 percent of suicide victims are male.
suicides_df = clean_intent_gender_data[clean_intent_gender_data['Intent'] == 'Suicide']
suicides_count = suicides_df['Deaths'].sum()

male_suicides_df = suicides_df[suicides_df['Gender'] == 'Male']
male_suicides_count = male_suicides_df['Deaths'].sum()

male_suicides_count / suicides_count
```

```python
# Verify: Women constitue only 15 percent of the total homicide victims.
homicides_df = clean_intent_gender_data[clean_intent_gender_data['Intent'] == 'Homicide']
homicides_count = homicides_df['Deaths'].sum()

female_homicides_df = homicides_df[homicides_df['Gender'] == 'Female']
female_homicides_df = female_homicides_df['Deaths'].sum()

female_homicides_df / homicides_count
```

```python
# Clean intent, gender, age, race data
clean_data = data[~data[['Intent', 'Gender', 'Age', 'Race']].eq('None selected').any(axis=1)]
clean_data.head()
```

```python
# Verify: Around two-thirds of homicide victims who are males in the age-group of 15--34 are black.
filtered_df = clean_data[(clean_data['Intent'] == 'Homicide') & (clean_data['Gender'] == 'Male') & (clean_data['Age'] == '15 - 34')]
filtered_death_count = filtered_df['Deaths'].sum()

black_filtered_df = filtered_df[filtered_df['Race'] == 'Black']
back_filtered_death_count = black_filtered_df['Deaths'].sum()

back_filtered_death_count / filtered_death_count
```
